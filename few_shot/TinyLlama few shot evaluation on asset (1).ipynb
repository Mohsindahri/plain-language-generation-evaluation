```python
# ================================================================================
# 📊 TINYLLAMA — EVALUATION METRICS FUNCTION
# SAME METRICS AS PHI PIPELINE
# ================================================================================

import evaluate
import textstat
import numpy as np
import re

print("\n" + "=" * 100)
print("🔧 LOADING EVALUATION METRICS")
print("=" * 100)

# Load metrics
sari_metric = evaluate.load("sari")
bleu_metric = evaluate.load("bleu")
rouge_metric = evaluate.load("rouge")
bertscore_metric = evaluate.load("bertscore")

print("✅ Metrics loaded")


# ================================================================================
# CLEAN PREDICTION
# ================================================================================

def clean_prediction(prediction):

    if prediction is None:
        return ""

    prediction = str(prediction)

    # Remove special tokens
    prediction = prediction.replace("<|endoftext|>", "")
    prediction = prediction.replace("<|assistant|>", "")
    prediction = prediction.replace("<|user|>", "")
    prediction = prediction.replace("<s>", "")
    prediction = prediction.replace("</s>", "")

    # If model accidentally returns prompt-style text,
    # keep only the Simplified portion
    if "Simplified :" in prediction:
        prediction = prediction.split("Simplified :", 1)[1]

    elif "Simplified:" in prediction:
        prediction = prediction.split("Simplified:", 1)[1]

    # Remove anything after accidental next-turn markers
    for marker in ["<|", "Original :", "Original:"]:
        if marker in prediction:
            prediction = prediction.split(marker, 1)[0]

    # Normalize whitespace
    prediction = re.sub(r"\s+", " ", prediction).strip()

    # Fix spaces before punctuation
    prediction = re.sub(r"\s+([,.!?;:])", r"\1", prediction)

    # Remove accidental spaces around brackets
    prediction = prediction.replace("( ", "(")
    prediction = prediction.replace(" )", ")")

    return prediction.strip()


# ================================================================================
# METRIC FUNCTION
# ================================================================================

def evaluate_one_sample(source, prediction, references):

    prediction = clean_prediction(prediction)

    # Safety
    if not prediction:
        prediction = ""

    references = [str(x).strip() for x in references]

    # --------------------------------------------------------------------------
    # SARI
    # --------------------------------------------------------------------------

    sari_result = sari_metric.compute(
        sources=[source],
        predictions=[prediction],
        references=[references]
    )

    sari_score = float(sari_result["sari"])

    # --------------------------------------------------------------------------
    # BLEU
    # --------------------------------------------------------------------------

    bleu_result = bleu_metric.compute(
        predictions=[prediction],
        references=[references]
    )

    bleu_score = float(bleu_result["bleu"])

    # --------------------------------------------------------------------------
    # ROUGE-L
    # --------------------------------------------------------------------------

    rouge_result = rouge_metric.compute(
        predictions=[prediction],
        references=[references[0]]
    )

    rouge_l = float(rouge_result["rougeL"])

    # --------------------------------------------------------------------------
    # BERTScore
    # --------------------------------------------------------------------------

    bert_result = bertscore_metric.compute(
        predictions=[prediction],
        references=[references[0]],
        lang="en"
    )

    bertscore_f1 = float(bert_result["f1"][0])

    # --------------------------------------------------------------------------
    # FKGL
    # --------------------------------------------------------------------------

    fkgl = float(
        textstat.flesch_kincaid_grade(prediction)
        if prediction.strip()
        else 0.0
    )

    return {
        "sari": sari_score,
        "bleu": bleu_score,
        "rougeL": rouge_l,
        "bertscore_f1": bertscore_f1,
        "fkgl": fkgl
    }


print("✅ evaluate_one_sample() defined")
print("✅ clean_prediction() defined")

print("\n" + "=" * 100)
print("✅ METRIC FUNCTION READY")
print("=" * 100)
```

```python
# ==================================================================================
# 🚀 TINYLLAMA — FULL FEW-SHOT ASSET EVALUATION
# 🔒 SAFE VERSION — DOES NOT CRASH ON EMPTY PREDICTIONS
# ==================================================================================

import os
import gc
import time
import numpy as np
import pandas as pd
import torch

# ==================================================================================
# CONFIGURATION
# ==================================================================================

OUTPUT_FILE = "TINYLLAMA_FEW_SHOT_ASSET_2359.csv"
CHECKPOINT_EVERY = 25
MAX_NEW_TOKENS = 64

print("=" * 100)
print("🚀 TINYLLAMA — FULL FEW-SHOT ASSET EVALUATION")
print("=" * 100)

print(f"Total samples : {len(originals)}")
print(f"Output file   : {OUTPUT_FILE}")
print(f"Checkpoint    : every {CHECKPOINT_EVERY} samples")
print(f"Max new tokens: {MAX_NEW_TOKENS}")

# ==================================================================================
# VERIFY REQUIRED OBJECTS
# ==================================================================================

assert model is not None
print("\n✅ Model found")

assert tokenizer is not None
print("✅ Tokenizer found")

assert len(originals) == len(reference_groups)
print("✅ ASSET dataset found")

assert callable(evaluate_one_sample)
print("✅ Metric function found")

# ==================================================================================
# HELPER — SAFE METRIC CALCULATION
# ==================================================================================

def safe_evaluate_one_sample(source, prediction, references):

    # --------------------------------------------------------------------------
    # EMPTY PREDICTION
    # --------------------------------------------------------------------------

    if prediction is None or not str(prediction).strip():

        print("\n⚠️ EMPTY PREDICTION")
        print("Assigning zero scores and continuing.")

        return {
            "sari": 0.0,
            "bleu": 0.0,
            "rougeL": 0.0,
            "bertscore_f1": 0.0,
            "fkgl": 0.0,
        }

    # --------------------------------------------------------------------------
    # NORMAL METRIC CALCULATION
    # --------------------------------------------------------------------------

    try:

        return evaluate_one_sample(
            source,
            prediction,
            references
        )

    except ZeroDivisionError:

        print("\n⚠️ BLEU ZERO-DIVISION ERROR")
        print("Assigning BLEU = 0.0 and continuing.")

        # Calculate the other metrics separately if possible.
        scores = {
            "sari": 0.0,
            "bleu": 0.0,
            "rougeL": 0.0,
            "bertscore_f1": 0.0,
            "fkgl": 0.0,
        }

        # SARI
        try:
            sari_result = sari_metric.compute(
                sources=[source],
                predictions=[prediction],
                references=[references]
            )
            scores["sari"] = float(sari_result["sari"])
        except Exception:
            pass

        # ROUGE-L
        try:
            rouge_result = rouge_metric.compute(
                predictions=[prediction],
                references=[references],
                use_stemmer=True
            )
            scores["rougeL"] = float(rouge_result["rougeL"])
        except Exception:
            pass

        # BERTScore
        try:
            bert_result = bertscore_metric.compute(
                predictions=[prediction],
                references=[references[0]],
                lang="en"
            )
            scores["bertscore_f1"] = float(
                np.mean(bert_result["f1"])
            )
        except Exception:
            pass

        # FKGL
        try:
            scores["fkgl"] = float(
                textstat.flesch_kincaid_grade(prediction)
            )
        except Exception:
            pass

        return scores

    except Exception as e:

        print("\n⚠️ METRIC ERROR")
        print(str(e))
        print("Assigning safe zero values and continuing.")

        return {
            "sari": 0.0,
            "bleu": 0.0,
            "rougeL": 0.0,
            "bertscore_f1": 0.0,
            "fkgl": 0.0,
        }


# ==================================================================================
# LOAD EXISTING CSV
# ==================================================================================

if os.path.exists(OUTPUT_FILE):

    print("\n🔄 Existing CSV found")

    results_df = pd.read_csv(OUTPUT_FILE)

    completed_ids = set(
        results_df["sample_id"].astype(int).tolist()
    )

    print(f"Completed samples: {len(completed_ids)}")

else:

    print("\n🆕 Starting new evaluation")

    results_df = pd.DataFrame()

    completed_ids = set()


# ==================================================================================
# MAIN EVALUATION LOOP
# ==================================================================================

total_samples = len(originals)

remaining = total_samples - len(completed_ids)

print(f"Remaining samples: {remaining}")

for idx in range(total_samples):

    # --------------------------------------------------------------------------
    # SKIP ALREADY COMPLETED SAMPLES
    # --------------------------------------------------------------------------

    if idx in completed_ids:
        continue

    sample_number = idx + 1

    source = originals[idx]
    references = reference_groups[idx]

    print("\n" + "=" * 100)
    print(
        f"🧪 TINYLLAMA FEW-SHOT — SAMPLE "
        f"{sample_number}/{total_samples}"
    )
    print("=" * 100)

    print("\n🔵 ORIGINAL")
    print("-" * 100)
    print(source)

    # ==========================================================================
    # CREATE FEW-SHOT PROMPT
    # ==========================================================================

    prompt = f"""Simplify the following sentences using clear and easy-to-understand language.

Preserve the original meaning.
Do not add explanations or commentary.
Return only the simplified sentence.

Here are some examples:

Original: The spacecraft consists of two main elements: the NASA Cassini orbiter and the ESA Huygens probe.
Simplified: The spacecraft has two main parts: the Cassini orbiter and the Huygens probe.

Original: Jeddah is the principal gateway to Mecca, Islam's holiest city.
Simplified: Jeddah is the main entrance to Mecca, Islam's holiest city.

Original: His next work follows an especially eventful day in the life of a successful neurosurgeon.
Simplified: His next work is about a very busy day in the life of a successful brain surgeon.

Now simplify this sentence:

Original: {source}
Simplified:"""

    # ==========================================================================
    # CHAT TEMPLATE
    # ==========================================================================

    messages = [
        {
            "role": "user",
            "content": prompt
        }
    ]

    formatted_input = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    # ==========================================================================
    # TOKENIZATION
    # ==========================================================================

    inputs = tokenizer(
        formatted_input,
        return_tensors="pt",
        truncation=True,
        max_length=2048
    )

    inputs = {
        k: v.to(model.device)
        for k, v in inputs.items()
    }

    input_length = inputs["input_ids"].shape[1]

    # ==========================================================================
    # VRAM RESET
    # ==========================================================================

    if torch.cuda.is_available():

        torch.cuda.empty_cache()
        torch.cuda.reset_peak_memory_stats()

    # ==========================================================================
    # GENERATION
    # ==========================================================================

    print("\n🤖 GENERATING...")

    start_time = time.perf_counter()

    try:

        with torch.inference_mode():

            output_ids = model.generate(
                **inputs,
                max_new_tokens=MAX_NEW_TOKENS,
                do_sample=False,
                num_beams=1,
                temperature=None,
                top_p=None,
                pad_token_id=tokenizer.pad_token_id,
                eos_token_id=tokenizer.eos_token_id,
                use_cache=True
            )

        inference_time = time.perf_counter() - start_time

        # ----------------------------------------------------------------------
        # ONLY TAKE NEW TOKENS
        # ----------------------------------------------------------------------

        generated_ids = output_ids[0][input_length:]

        prediction = tokenizer.decode(
            generated_ids,
            skip_special_tokens=True
        )

        prediction = prediction.strip()

    except Exception as e:

        inference_time = time.perf_counter() - start_time

        print("\n❌ GENERATION ERROR")
        print(str(e))

        prediction = ""

    # ==========================================================================
    # PEAK VRAM
    # ==========================================================================

    if torch.cuda.is_available():

        peak_vram = (
            torch.cuda.max_memory_allocated()
            / (1024 ** 3)
        )

    else:

        peak_vram = 0.0

    # ==========================================================================
    # CLEAN PREDICTION
    # ==========================================================================

    try:

        prediction = clean_prediction(
            prediction
        )

    except Exception as e:

        print("\n⚠️ CLEANING ERROR")
        print(str(e))

        prediction = prediction.strip()


    # ==========================================================================
    # DISPLAY PREDICTION
    # ==========================================================================

    print("\n🤖 PREDICTION")
    print("-" * 100)

    if prediction:

        print(prediction)

    else:

        print("[EMPTY PREDICTION]")


    # ==========================================================================
    # METRICS — SAFE
    # ==========================================================================

    print("\n📊 METRICS")
    print("-" * 100)

    scores = safe_evaluate_one_sample(
        source,
        prediction,
        references
    )

    sari_score = float(
        scores.get("sari", 0.0)
    )

    bleu_score = float(
        scores.get("bleu", 0.0)
    )

    rouge_score = float(
        scores.get("rougeL", 0.0)
    )

    bert_score = float(
        scores.get("bertscore_f1", 0.0)
    )

    fkgl_score = float(
        scores.get("fkgl", 0.0)
    )

    print(f"SARI          : {sari_score:.4f}")
    print(f"BLEU          : {bleu_score:.4f}")
    print(f"ROUGE-L       : {rouge_score:.4f}")
    print(f"BERTScore-F1  : {bert_score:.4f}")
    print(f"FKGL          : {fkgl_score:.4f}")


    # ==========================================================================
    # PERFORMANCE
    # ==========================================================================

    print("\n⚡ PERFORMANCE")
    print("-" * 100)

    print(
        f"Inference time: {inference_time:.3f} sec"
    )

    print(
        f"Peak VRAM     : {peak_vram:.3f} GB"
    )


    # ==========================================================================
    # SAVE RESULT
    # ==========================================================================

    row = {
        "sample_id": sample_number,
        "source": source,
        "prediction": prediction,

        "references": " ||| ".join(references),

        "sari": sari_score,
        "bleu": bleu_score,
        "rougeL": rouge_score,
        "bertscore_f1": bert_score,
        "fkgl": fkgl_score,

        "inference_time_sec": inference_time,
        "peak_vram_gb": peak_vram
    }

    results_df = pd.concat(
        [
            results_df,
            pd.DataFrame([row])
        ],
        ignore_index=True
    )

    # ==========================================================================
    # CHECKPOINT SAVE — EVERY 25 SAMPLES
    # ==========================================================================

    completed_now = len(results_df)

    if completed_now % CHECKPOINT_EVERY == 0:

        results_df.to_csv(
            OUTPUT_FILE,
            index=False,
            encoding="utf-8-sig"
        )

        print("\n" + "=" * 100)
        print("💾 CHECKPOINT SAVED")
        print("=" * 100)
        print(f"Completed : {completed_now}/{total_samples}")
        print(f"Remaining : {total_samples - completed_now}")
        print(f"File      : {OUTPUT_FILE}")
        print("=" * 100)

    # ==========================================================================
    # MEMORY CLEANUP
    # ==========================================================================

    del inputs
    del output_ids

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    print("\n" + "=" * 100)
    print(f"✅ SAMPLE {sample_number} COMPLETE")
    print("=" * 100)


# ==================================================================================
# FINAL SAVE
# ==================================================================================

results_df.to_csv(
    OUTPUT_FILE,
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 100)
print("🎉 TINYLLAMA FEW-SHOT EVALUATION COMPLETE")
print("=" * 100)

print(f"Total samples : {len(results_df)}")
print(f"Output file   : {OUTPUT_FILE}")

print("=" * 100)
```

```python
# ==================================================================================
# 📊 TINYLLAMA FEW-SHOT — DESCRIPTIVE STATISTICS
# Mean | Median | Min | Max | Std
# ==================================================================================

import pandas as pd
import numpy as np
import os

OUTPUT_FILE = "TINYLLAMA_FEW_SHOT_ASSET_2359.csv"

print("=" * 100)
print("📂 LOADING TINYLLAMA FEW-SHOT RESULTS")
print("=" * 100)

# ------------------------------------------------------------------------------
# LOAD CSV
# ------------------------------------------------------------------------------

if not os.path.exists(OUTPUT_FILE):
    raise FileNotFoundError(
        f"❌ File not found: {OUTPUT_FILE}"
    )

results_df = pd.read_csv(OUTPUT_FILE)

print(f"✅ CSV loaded successfully")
print(f"Rows in CSV: {len(results_df)}")

print("\nColumns:")
for col in results_df.columns:
    print(f" - {col}")


# ==================================================================================
# 🔍 CHECK FOR DUPLICATES
# ==================================================================================

print("\n" + "=" * 100)
print("🔍 CHECKING DUPLICATES")
print("=" * 100)

if "sample_index" in results_df.columns:

    duplicate_count = results_df.duplicated(
        subset=["sample_index"]
    ).sum()

    print(f"Duplicate sample_index rows: {duplicate_count}")

    if duplicate_count > 0:

        print("⚠️ Removing duplicate sample_index rows...")

        results_df = (
            results_df
            .drop_duplicates(
                subset=["sample_index"],
                keep="first"
            )
            .sort_values("sample_index")
            .reset_index(drop=True)
        )

        print(f"Rows after cleaning: {len(results_df)}")

        # Save cleaned version
        results_df.to_csv(
            OUTPUT_FILE,
            index=False,
            encoding="utf-8-sig"
        )

        print("💾 Cleaned CSV saved.")

else:
    print("⚠️ sample_index column not found.")
    print("Statistics will use all rows.")


# ==================================================================================
# 📊 METRIC COLUMNS
# ==================================================================================

metric_columns = [
    "sari",
    "bleu",
    "rougeL",
    "bertscore_f1",
    "fkgl",
    "inference_time_sec",
    "peak_vram_gb"
]

# Keep only columns that actually exist
available_metrics = [
    col for col in metric_columns
    if col in results_df.columns
]

print("\n" + "=" * 100)
print("📊 AVAILABLE METRICS")
print("=" * 100)

for col in available_metrics:
    print(f"✅ {col}")


# ==================================================================================
# 📈 CALCULATE DESCRIPTIVE STATISTICS
# ==================================================================================

statistics_rows = []

for metric in available_metrics:

    values = pd.to_numeric(
        results_df[metric],
        errors="coerce"
    ).dropna()

    statistics_rows.append({
        "Metric": metric,
        "Mean": values.mean(),
        "Median": values.median(),
        "Min": values.min(),
        "Max": values.max(),
        "Std": values.std()
    })


stats_df = pd.DataFrame(statistics_rows)


# ==================================================================================
# 🎯 ROUND RESULTS
# ==================================================================================

stats_df = stats_df.round(4)


# ==================================================================================
# 🖥️ DISPLAY RESULTS
# ==================================================================================

print("\n" + "=" * 100)
print("📊 TINYLLAMA FEW-SHOT — DESCRIPTIVE STATISTICS")
print("=" * 100)

print(stats_df.to_string(index=False))


# ==================================================================================
# 💾 SAVE STATISTICS
# ==================================================================================

STATS_FILE = "TINYLLAMA_FEW_SHOT_STATISTICS.csv"

stats_df.to_csv(
    STATS_FILE,
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 100)
print("🎉 STATISTICS COMPLETE")
print("=" * 100)

print(f"Samples analyzed : {len(results_df)}")
print(f"Statistics file  : {STATS_FILE}")

print("=" * 100)
```
